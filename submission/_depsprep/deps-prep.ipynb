```python

import subprocess, sys, os, shutil
# Resolve the FULL transformers-main dependency tree WITH internet into a target dir.
subprocess.run([sys.executable,"-m","pip","install","--target=/kaggle/working/deps","-U",
    "git+https://github.com/huggingface/transformers.git","peft","trl","accelerate","datasets","compressed-tensors","torchao>=0.17.0"], check=True)
# Drop torch/nvidia/triton so the OFFLINE kernel keeps the image's Blackwell torch (deps stays self-consistent otherwise)
for d in list(os.listdir("/kaggle/working/deps")):
    if d.split("-")[0].lower() in ("torch","torchvision","torchaudio","triton") or d.startswith(("nvidia","cuda")):
        shutil.rmtree(os.path.join("/kaggle/working/deps",d), ignore_errors=True)
sz=sum(os.path.getsize(os.path.join(dp,f)) for dp,_,fs in os.walk("/kaggle/working/deps") for f in fs)/1e6
print(f"deps bundle ready: {sz:.0f} MB")
import transformers; print("bundled transformers", transformers.__version__)
```
